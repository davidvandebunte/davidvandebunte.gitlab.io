### 1.3.1. Bayesian data analysis 

See [Bayesian Data Analysis](http://www.stat.columbia.edu/~gelman/book/) (BDA) for the book that coined this term. The author has already referred to Gelman's material in footnotes and will continue to do so.

Compare [Probability interpretations](https://en.wikipedia.org/wiki/Probability_interpretations), which generally speaking provides a much clearer introduction to this topic. The author is not referring to [Resampling (statistics)](https://en.wikipedia.org/wiki/Resampling_(statistics)) in this section, somewhat confusingly. The author only uses the words [Sampling distribution](https://en.wikipedia.org/wiki/Sampling_distribution) without providing any kind of definition or background.

The author defines [Bayesianism](https://en.wikipedia.org/w/index.php?title=Bayesianism&redirect=no) as a normative description of rational belief, though it's not clear that this is how the word is always used on e.g. Wikipedia (it appears to often simply mean Bayesian statistics).

The author's final point (that BDA is more intuitive) is among the strongest for it. This may be because the Bayesian interpretation is simpler than the frequentist interpretation. Compare the length of the the descriptions in [Credible interval](https://en.wikipedia.org/wiki/Credible_interval) and [Confidence interval](https://en.wikipedia.org/wiki/Confidence_interval); although the latter article does a great job of boiling down what a confidence interval is (especially with the drawing [Normal distribution 50% CI illustration](https://en.wikipedia.org/wiki/Confidence_interval#/media/File:Normal_distribution_50%25_CI_illustration.svg)) the result is just not as succinct. If someone can skip past all the misuses of [p-value](https://en.wikipedia.org/wiki/P-value) and simply never need to deal with them, their life could be much easier (unlikely for most people in our generation).

Another strong argument for the Bayesian perspective is that the "data products" (referred to as "statistical propositions" in [Statistical inference § Introduction](https://en.wikipedia.org/wiki/Statistical_inference#Introduction), and elsewhere as [probabilistic propositions](https://en.wikipedia.org/wiki/Probabilistic_proposition)) are richer. Clearly [Highest posterior density interval](https://en.wikipedia.org/wiki/Credible_interval#/media/File:Highest_posterior_density_interval.svg) lets you conclude something about where the parameter sits if it isn't in the 95% interval (it's likely higher) when the same can't be said with a confidence interval as a result.
