```r
?cars
```

Fit a linear regression of distance on speed:

```r
m <- lm(dist ~ speed, data=cars)
m
```

Estimated coefficients from the model:

```r
coef(m)
```

Plot residuals against speed:

```r
plot(resid(m) ~ speed, data=cars)
```
